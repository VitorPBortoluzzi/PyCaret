```python
from pycaret.classification import *
```

```python
clf = setup(
    data=df,
    # target="Churn",
    target = "Risco_Internacao",
    session_id=123
)
```

```python
best_model = compare_models()
```

```python
knn = create_model(best_model)
```

```python
evaluate_model(knn)
```

```python
tuned_knn = tune_model(knn)
```

```python
print(knn)
print(tuned_knn)
```

```python
predictions = predict_model(tuned_knn)
```
